1.Apa perbedaan supervised, unsupervised, dan reinforcement learning?


Perbedaan supervised, unsupervised, dan reinforcement learning terletak pada cara pembelajarannya. Supervised adalah belajar dari data yang sudah ada jawaban atau label, memiliki dua jenis yaitu klasifikasi dan regresi. Tujuannya untuk memprediksi. Contohnya seperti filter spam dan dianalogikan sebagai kunci jawaban. Sedangkan unsupervised adalah belajar dari data yang tidak ada jawaban atau label, kemudian mencari data yang meliliki kemiripan atau yang paling dekat. Reinforcement belajar dengan cara dilatih kalo bener dapat point kalo salah dapat punishment
